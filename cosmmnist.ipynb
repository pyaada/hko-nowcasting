```python
from data.dutils import CoSMMNIST
data_root = 'data'
example_set = CoSMMNIST(root=data_root, is_train=True, n_frames_input=10, n_frames_output=10, num_objects=[4])
```

```python
example_set.__len__()
```

```python
sample_context_video_in, sample_context_video_out, sample_video_in, sample_video_out = example_set.__getitem__(0)
print(sample_context_video_in.shape, sample_context_video_out.shape, sample_video_in.shape, sample_video_out.shape)
```

```python
import matplotlib.pyplot as plt

# Get image sizes
context_in_size = sample_context_video_in.shape[2:]
context_out_size = sample_context_video_out.shape[2:]
crop_in_size = sample_video_in.shape[2:]
crop_out_size = sample_video_out.shape[2:]

# Create figure with adjusted subplot sizes
fig = plt.figure(figsize=(20, 8))
gs = fig.add_gridspec(4, 10, height_ratios=[context_in_size[0], context_out_size[0], 
                                          crop_in_size[0], crop_out_size[0]])

# Plot input context frames with original size
for i in range(10):
    ax = fig.add_subplot(gs[0, i])
    ax.imshow(sample_context_video_in[i, 0], cmap='gray', extent=[0, context_in_size[1], context_in_size[0], 0])
    ax.axis('off')
    if i == 0:
        ax.set_title('Input Context Frames')

# Plot output context frames with original size
for i in range(10):
    ax = fig.add_subplot(gs[1, i])
    ax.imshow(sample_context_video_out[i, 0], cmap='gray', extent=[0, context_out_size[1], context_out_size[0], 0])
    ax.axis('off')
    if i == 0:
        ax.set_title('Output Context Frames')

# Plot cropped input frames with their size
for i in range(10):
    ax = fig.add_subplot(gs[2, i])
    ax.imshow(sample_video_in[i, 0], cmap='gray', extent=[0, crop_in_size[1], crop_in_size[0], 0])
    ax.axis('off')
    if i == 0:
        ax.set_title('Input Frames')

# Plot cropped output frames with their size
for i in range(10):
    ax = fig.add_subplot(gs[3, i])
    ax.imshow(sample_video_out[i, 0], cmap='gray', extent=[0, crop_out_size[1], crop_out_size[0], 0])
    ax.axis('off')
    if i == 0:
        ax.set_title('Output Frames')

plt.tight_layout()
plt.show()
```
